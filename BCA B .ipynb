### Recusion

- when a function call itself again and again untill given condition is not achived or false.

```python
# wpp to print 10 natural number

def natural(n):
    if n==11:
        return
    else:
        print(n)
        natural(n+1)
```

```python
natural(1)
```

```python
## wpp to print 10 natural number in reverse order.

def natural(n):
    if n==0:
        return
    else:
        print(n)
        natural(n-1)
```

```python
natural(10)
```

```python
# wpp to create factorial


def fact(n):
    if n==0:
        return 1
    else:
        return fact(n-1)*n
```

```python
class ATM:
    def __init__(self,Amount):
        self.__Amount = Amount
    def check_balance(self):
        return f"your current Amount is {self.__Amount}."
    def deposite(self,Amount):
        if Amount>0:
            self.__Amount+=Amount
            print(f"your amount {Amount} Deposite Successfully.")
            return f"your current Amount is {self.__Amount}."
        else:
            print("please enter valid amount.")
    def withdraw(self,Amount):
        if Amount>0 and Amount<=self.__Amount:
            self.__Amount-=Amount
            print(f"your amount {Amount} Withdraw Successfully.")
            return f"your current Amount is {self.__Amount}."
        else:
            print("insufficient Amount in your Account.")
```

```python
atm = ATM(200)
```

```python
atm.check_balance()
```

```python
atm.deposite(20)
```

```python
atm.withdraw(20)
```

```python
atm.withdraw(500)
```

```python
atm.deposite(-500)
```
